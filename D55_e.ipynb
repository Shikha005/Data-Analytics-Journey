```python
data = [[1, 4, 6], [9, 2, 8], [3, 7, 5]]
target = 10
found_pair = None

for i in range(len(data)):
    for j in range(len(data[i])):
        for k in range(j + 1, len(data[i])):
            if data[i][j] + data[i][k] == target:
                found_pair = (data[i][j], data[i][k], f"row {i}")
                break
        if found_pair:
            break
    if found_pair:
        break

print("Pair found:", found_pair)
```

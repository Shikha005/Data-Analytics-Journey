```python
def run_length_encode(text):
    encoded = ""
    i = 0
    while i < len(text):
        count = 1
        while i + 1 < len(text) and text[i] == text[i + 1]:
            i += 1
            count += 1
        encoded += text[i] + str(count)
        i += 1
    return encoded

print(run_length_encode("aaabbbcccd"))  # a3b3c3d1
```

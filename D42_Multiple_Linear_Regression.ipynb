```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
import pandas as pd
import numpy as np

np.random.seed(42)
n = 150

data = {
    "experience":  np.random.randint(0, 15, n),
    "skills":      np.random.randint(1, 10, n),
    "education":   np.random.randint(1, 4, n),
    "interviews":  np.random.randint(1, 10, n),
    "salary":      np.random.randint(20000, 150000, n)
}

df = pd.DataFrame(data)

df["salary"] = (df["experience"] * 5000 +
                df["skills"] * 3000 +
                df["education"] * 8000 +
                np.random.randint(-5000, 5000, n) + 15000)

x = df.drop("salary", axis=1)
y = df["salary"]

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

model = LinearRegression()
model.fit(x_train, y_train)

y_pred = model.predict(x_test)

print("multiple linear regression - salary prediction")
print("-" * 45)
print("r2 score:", round(r2_score(y_test, y_pred), 3))
print("-" * 45)
print("feature coefficients:")
for feat, coef in zip(x.columns, model.coef_):
    print(" ", feat, "->", round(coef, 2))
print("intercept:", round(model.intercept_, 2))

new_emp = [[5, 7, 2, 4]]
pred_salary = model.predict(new_emp)
print("-" * 45)
print("predicted salary:", round(pred_salary[0], 0))
```

```python
import numpy as np
from scipy import stats

data = np.array([50, 60, 65, 70, 72, 75, 80, 85, 90, 95])

mean = np.mean(data)
std_dev = np.std(data)

z_scores = stats.zscore(data)

print("Data:", data)
print("Mean:", mean)
print("Standard Deviation:", std_dev)
print("Z-Scores:", z_scores)

# Probability of a value being less than 80 in this distribution
prob = stats.norm.cdf(80, loc=mean, scale=std_dev)
print("Probability (X < 80):", prob)
```

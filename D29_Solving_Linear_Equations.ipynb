```python
import numpy as np


# Coefficient matrix
A = np.array([
    [2,3],
    [5,4]
])


# Constant matrix
B = np.array([8,13])


# Solve Ax = B

solution = np.linalg.solve(A,B)


print("Solution:")
print(solution)


print("\nValue of x:")
print(solution[0])


print("\nValue of y:")
print(solution[1])
```

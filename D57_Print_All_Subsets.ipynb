```python
def power_set(arr):
    n = len(arr)
    total = 2 ** n
    result = []

    for i in range(total):
        subset = []
        for j in range(n):
            if i & (1 << j):
                subset.append(arr[j])
        result.append(subset)

    return result

items = ['a', 'b', 'c']
for s in power_set(items):
    print(s)
```

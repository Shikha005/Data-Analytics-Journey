```python
def calculate_range(data):
    return max(data) - min(data)


def frequency_distribution(data):
    frequency = {}
    for num in data:
        frequency[num] = frequency.get(num, 0) + 1
    return frequency


scores = [55, 60, 60, 70, 80, 80, 80, 90, 100, 55]

print("Data:", scores)
print("Range:", calculate_range(scores))

freq_dist = frequency_distribution(scores)
print("Frequency Distribution:")
for value, freq in freq_dist.items():
    print(f"  {value}: {freq} time(s)")
```

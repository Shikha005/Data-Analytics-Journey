```python
n = int(input("Enter a number: "))

largest = -1
second_largest = -1

while n > 0:
    digit = n % 10
    n //= 10

    if digit > largest:
        second_largest = largest
        largest = digit

    elif digit > second_largest and digit != largest:
        second_largest = digit

if second_largest == -1:
    print("Second largest digit does not exist")
else:
    print("Second largest digit:", second_largest)
```

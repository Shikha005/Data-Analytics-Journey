```python
import numpy as np
from scipy import stats

method_a = [85, 88, 90, 92, 87]
method_b = [78, 82, 80, 85, 79]
method_c = [92, 95, 91, 94, 96]

f_stat, p_value = stats.f_oneway(method_a, method_b, method_c)

print("F-Statistic:", f_stat)
print("P-Value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Result: At least one group mean is significantly different.")
else:
    print("Result: No significant difference between group means.")
```

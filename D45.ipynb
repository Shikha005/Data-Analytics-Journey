```python
import pandas as pd
import numpy as np

np.random.seed(1)
data = {
    'age': [25, np.nan, 30, 45, np.nan, 29, 41, np.nan, 33, 38],
    'salary': [50000, 60000, np.nan, 75000, 48000, np.nan, 82000, 51000, np.nan, 67000],
    'city': ['Delhi', 'Mumbai', np.nan, 'Chennai', 'Delhi', 'Pune', np.nan, 'Mumbai', 'Delhi', 'Chennai']
}
df = pd.DataFrame(data)

print("Original Data:\n", df)

missing_count = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100
missing_summary = pd.DataFrame({'Missing Count': missing_count, 'Missing %': missing_percent})
print("\nMissing Value Summary:\n", missing_summary)

for col in df.columns:
    if df[col].dtype in ['float64', 'int64']:
        df[col] = df[col].fillna(df[col].median())
    else:
        df[col] = df[col].fillna(df[col].mode()[0])

print("\nData after imputation:\n", df)

print("\nRemaining missing values:\n", df.isnull().sum())
```

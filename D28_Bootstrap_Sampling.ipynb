```python
import numpy as np

marks = np.array([60,65,70,72,75,78,80,82,85,90])

bootstrap_means = []

for i in range(1000):
    sample = np.random.choice(marks,
                              size=len(marks),
                              replace=True)
    bootstrap_means.append(np.mean(sample))

print("Bootstrap Mean:", np.mean(bootstrap_means))
print("Bootstrap Std:", np.std(bootstrap_means))
```

```python
class Car:
    def calculate_fare(self, distance):
        return distance * 15


class Bike:
    def calculate_fare(self, distance):
        return distance * 8


class Bus:
    def calculate_fare(self, distance):
        return distance * 5


def show_fare(vehicle, distance):
    print("Fare:", vehicle.calculate_fare(distance))


car = Car()
bike = Bike()
bus = Bus()

show_fare(car, 10)
show_fare(bike, 10)
show_fare(bus, 10)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, cross_val_score
import numpy as np

cancer = load_breast_cancer()
x = cancer.data
y = cancer.target

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

print("original features:", x.shape[1])

pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("pca",    PCA(n_components=10)),
    ("model",  RandomForestClassifier(n_estimators=100, random_state=42))
])

pipe.fit(x_train, y_train)

print("after pca features:", 10)
print("-" * 40)
print("train accuracy:", round(pipe.score(x_train, y_train), 3))
print("test accuracy:", round(pipe.score(x_test, y_test), 3))

cv = cross_val_score(pipe, x, y, cv=5)
print("cv accuracy:", round(np.mean(cv), 3))
print("-" * 40)

pca_step = pipe.named_steps["pca"]
print("variance explained:",
      round(sum(pca_step.explained_variance_ratio_) * 100, 1), "%")
```

```python
class MathOperations:
    school_name = "ABC Public School"

    def __init__(self, a, b):
        self.a = a
        self.b = b

    def add(self):
        return self.a + self.b

    @staticmethod
    def multiply(x, y):
        return x * y

    @classmethod
    def get_school_name(cls):
        return cls.school_name


obj = MathOperations(5, 10)
print("Sum:", obj.add())
print("Product:", MathOperations.multiply(4, 6))
print("School:", MathOperations.get_school_name())
```

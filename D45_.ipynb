```python
import pandas as pd
import numpy as np
from scipy import stats

np.random.seed(2)
income = np.random.normal(50000, 10000, 95).tolist() + [200000, 5000, 210000, 3000, 195000]
df = pd.DataFrame({'income': income})

print("Total records:", len(df))
print(df['income'].describe())

Q1 = df['income'].quantile(0.25)
Q3 = df['income'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

iqr_outliers = df[(df['income'] < lower_bound) | (df['income'] > upper_bound)]
print(f"\nIQR bounds: [{lower_bound:.2f}, {upper_bound:.2f}]")
print("IQR outliers found:", len(iqr_outliers))
print(iqr_outliers)

df['z_score'] = np.abs(stats.zscore(df['income']))
z_outliers = df[df['z_score'] > 3]
print("\nZ-score outliers found:", len(z_outliers))
print(z_outliers)

df['income_capped'] = np.where(df['income'] > upper_bound, upper_bound,
                         np.where(df['income'] < lower_bound, lower_bound, df['income']))

print("\nBefore vs After capping (describe):")
print(df[['income', 'income_capped']].describe())
```

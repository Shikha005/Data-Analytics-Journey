```python
def tower_of_hanoi_iterative(n):
    total_moves = 2 ** n - 1
    towers = {'A': list(range(n, 0, -1)), 'B': [], 'C': []}

    source, auxiliary, target = ('A', 'B', 'C') if n % 2 == 0 else ('A', 'C', 'B')

    for move in range(1, total_moves + 1):
        if move % 3 == 1:
            src, dst = source, target
        elif move % 3 == 2:
            src, dst = source, auxiliary
        else:
            src, dst = auxiliary, target

        if not towers[dst] or (towers[src] and towers[src][-1] < towers[dst][-1]):
            disk = towers[src].pop()
            towers[dst].append(disk)
        else:
            disk = towers[dst].pop()
            towers[src].append(disk)

        print(f"Move {move}: {towers}")

tower_of_hanoi_iterative(3)
```

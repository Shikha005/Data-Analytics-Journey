```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

cancer = load_breast_cancer()
x = cancer.data
y = cancer.target

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

dt = DecisionTreeClassifier(random_state=42)
rf = RandomForestClassifier(n_estimators=100, random_state=42)

dt.fit(x_train, y_train)
rf.fit(x_train, y_train)

print("decision tree vs random forest")
print("-" * 35)
print("decision tree:")
print("  train:", round(dt.score(x_train, y_train), 3))
print("  test: ", round(dt.score(x_test, y_test), 3))
print("-" * 35)
print("random forest:")
print("  train:", round(rf.score(x_train, y_train), 3))
print("  test: ", round(rf.score(x_test, y_test), 3))
```

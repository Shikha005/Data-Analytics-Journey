```python
n = int(input("Enter a number: "))

original = n
digit = 0

while digit <= 9:
    temp = original
    count = 0

    while temp > 0:
        current_digit = temp % 10

        if current_digit == digit:
            count += 1

        temp //= 10

    if count > 0:
        print(digit, "→", count)

    digit += 1
```

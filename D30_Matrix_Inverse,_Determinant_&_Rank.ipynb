```python
import numpy as np


# Creating matrix
A = np.array([
    [4, 7],
    [2, 6]
])


print("Matrix A:")
print(A)


# Determinant
determinant = np.linalg.det(A)

print("\nDeterminant:")
print(determinant)


# Inverse
inverse = np.linalg.inv(A)

print("\nInverse of Matrix A:")
print(inverse)


# Rank
rank = np.linalg.matrix_rank(A)

print("\nRank of Matrix A:")
print(rank)
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary

    @property
    def salary(self):
        return self.__salary

    @salary.setter
    def salary(self, value):
        if value > 0:
            self.__salary = value
        else:
            print("Salary must be positive.")


emp = Employee("Sneha", 50000)
print("Salary:", emp.salary)

emp.salary = 60000
print("Updated Salary:", emp.salary)

emp.salary = -5000
```

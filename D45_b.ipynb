```python
import pandas as pd
import numpy as np

np.random.seed(4)
df = pd.DataFrame({
    'age': np.random.randint(18, 65, 200),
    'income': np.random.normal(55000, 15000, 200).round(2),
    'experience': np.random.randint(0, 40, 200),
    'score': np.random.uniform(0, 100, 200).round(2)
})

df = pd.concat([df, df.iloc[:5]], ignore_index=True)
df.loc[10, 'age'] = -5   # invalid value

print("Shape:", df.shape)
print("\nData types:\n", df.dtypes)

dup_count = df.duplicated().sum()
print(f"\nDuplicate rows found: {dup_count}")
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)

invalid_ages = df[df['age'] < 0]
print(f"\nInvalid age entries: {len(invalid_ages)}")
df = df[df['age'] >= 0]

print("\nSummary statistics:\n", df.describe())

print("\nCorrelation matrix:\n", df.corr().round(2))

print("\nSkewness:\n", df.skew(numeric_only=True))

for col in ['age', 'income', 'experience', 'score']:
    Q1, Q3 = df[col].quantile([0.25, 0.75])
    IQR = Q3 - Q1
    lower, upper = Q1 - 1.5*IQR, Q3 + 1.5*IQR
    count = ((df[col] < lower) | (df[col] > upper)).sum()
    print(f"{col}: {count} outliers")
```

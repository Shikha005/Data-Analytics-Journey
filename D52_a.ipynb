```python
n = int(input("Enter a number: "))

original = n
digits = 0

temp = n
while temp > 0:
    digits += 1
    temp //= 10

total = 0
temp = n

while temp > 0:
    digit = temp % 10
    total += digit ** digits
    temp //= 10

if total == original:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```

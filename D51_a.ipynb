```python
numbers = [3, 7, 12, 18, 22, 5, 9, 14]
target = 12
found = False

for idx, num in enumerate(numbers):
    if num % 2 != 0:
        continue  # skip odd numbers
    if num == target:
        print(f"Found {target} at index {idx}")
        found = True
        break

if not found:
    print("Number not found among even numbers")
```

```python
def is_armstrong(num):
    digits = str(num)
    power = len(digits)
    total = sum(int(d) ** power for d in digits)
    return total == num

start, end = 1, 1000
armstrong_numbers = [num for num in range(start, end + 1) if is_armstrong(num)]
print(armstrong_numbers)  # [1, 2, ..., 153, 370, 371, 407]
```

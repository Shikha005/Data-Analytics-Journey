```python
data = [1, 2, 2, 3, 3, 3, 1, 1, 1, 1, 5]

max_run = 1
current_run = 1
max_value = data[0]

for i in range(1, len(data)):
    if data[i] == data[i - 1]:
        current_run += 1
        if current_run > max_run:
            max_run = current_run
            max_value = data[i]
    else:
        current_run = 1

print(f"Longest run: {max_value} repeated {max_run} times")
```

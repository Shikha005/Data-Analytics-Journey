```python
import numpy as np


def gram_schmidt(vectors):

    orthogonal_vectors = []

    for v in vectors:

        u = v.copy().astype(float)

        for prev in orthogonal_vectors:
            projection = (
                np.dot(v, prev) /
                np.dot(prev, prev)
            ) * prev

            u = u - projection

        orthogonal_vectors.append(u)

    return orthogonal_vectors



# Input vectors

vectors = [
    np.array([1, 1, 0]),
    np.array([1, 0, 1]),
    np.array([0, 1, 1])
]


result = gram_schmidt(vectors)


print("Original Vectors:")
for v in vectors:
    print(v)


print("\nOrthogonal Vectors:")
for v in result:
    print(v)


print("\nNormalized Orthogonal Vectors:")

for v in result:
    print(v / np.linalg.norm(v))
```

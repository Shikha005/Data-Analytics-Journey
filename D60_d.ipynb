```python
def count_n_queens(n):
    def is_safe(board, row, col):
        for i in range(row):
            if board[i] == col or abs(board[i] - col) == abs(i - row):
                return False
        return True

    def solve(board, row, count):
        if row == n:
            return count + 1
        for col in range(n):
            if is_safe(board, row, col):
                board[row] = col
                count = solve(board, row + 1, count)
        return count

    board = [-1] * n
    return solve(board, 0, 0)

print(count_n_queens(8))  # 92 solutions
```

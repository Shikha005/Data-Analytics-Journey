```python
def has_cycle(sequence_func, start, limit=1000):
    slow = fast = start
    steps = 0

    while steps < limit:
        slow = sequence_func(slow)
        fast = sequence_func(sequence_func(fast))

        if slow == fast:
            return True
        steps += 1

    return False

# Example: sequence that cycles: 1 -> 2 -> 3 -> 4 -> 2 -> 3 -> 4 ...
def seq(x):
    mapping = {1: 2, 2: 3, 3: 4, 4: 2}
    return mapping[x]

print(has_cycle(seq, 1))  # True
```

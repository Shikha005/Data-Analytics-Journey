```python
rows, cols = 5, 10
result = [[0] * cols for _ in range(rows)]

col = 0
direction = 1  # 1 = down, -1 = up
row = 0

for col in range(cols):
    result[row][col] = 1
    if row == 0:
        direction = 1
    elif row == rows - 1:
        direction = -1
    row += direction

for r in result:
    print(" ".join(str(x) for x in r))
```

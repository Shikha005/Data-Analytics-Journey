```python
import pandas as pd
import numpy as np
from scipy.stats import zscore

salary = [25000,27000,28000,26000,25500,29000,31000,90000]

df = pd.DataFrame({"Salary": salary})

df["Z_Score"] = zscore(df["Salary"])

outliers = df[np.abs(df["Z_Score"]) > 2]

print(df)
print("\nOutliers:")
print(outliers)
```

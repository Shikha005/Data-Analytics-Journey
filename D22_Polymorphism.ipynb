```python
class Animal:
    def sound(self):
        print("some animal sound")

class Dog(Animal):
    def sound(self):
        print("dog says: woof woof!")

class Cat(Animal):
    def sound(self):
        print("cat says: meow!")

class Cow(Animal):
    def sound(self):
        print("cow says: moo!")

class Duck(Animal):
    def sound(self):
        print("duck says: quack quack!")

# same method, different output
animals = [Dog(), Cat(), Cow(), Duck()]

for animal in animals:
    animal.sound()
```

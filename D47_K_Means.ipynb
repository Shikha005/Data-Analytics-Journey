```python
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

X, _ = make_blobs(n_samples=300, centers=4, cluster_std=1.2, random_state=7)

inertias = []
sil_scores = []
K_range = range(2, 8)

for k in K_range:
    km = KMeans(n_clusters=k, random_state=7, n_init=10)
    labels = km.fit_predict(X)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X, labels))

print("k | Inertia | Silhouette Score")
for k, inertia, sil in zip(K_range, inertias, sil_scores):
    print(f"{k} | {inertia:.2f} | {sil:.4f}")

best_k = K_range[np.argmax(sil_scores)]
print(f"\nBest k based on silhouette score: {best_k}")

final_km = KMeans(n_clusters=best_k, random_state=7, n_init=10)
final_labels = final_km.fit_predict(X)
print("Cluster centers:\n", final_km.cluster_centers_)
```

```python
import numpy as np

X = np.array([1, 2, 3, 4, 5, 6, 7, 8])
y = np.array([2, 4, 5, 4, 5, 7, 8, 9])

X = (X - X.mean()) / X.std()
m, b = 0.0, 0.0
lr = 0.01
epochs = 1000
n = len(X)

for epoch in range(epochs):
    y_pred = m * X + b
    error = y_pred - y
    dm = (2/n) * np.dot(error, X)
    db = (2/n) * np.sum(error)
    m -= lr * dm
    b -= lr * db
    if epoch % 200 == 0:
        loss = np.mean(error ** 2)
        print(f"Epoch {epoch}: Loss={loss:.4f}, m={m:.4f}, b={b:.4f}")

print(f"\nFinal: y = {m:.3f}x + {b:.3f}")
```

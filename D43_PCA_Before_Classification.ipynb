```python
from sklearn.decomposition import PCA
from sklearn.svm import SVC
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
import time

digits = load_digits()
x = digits.data
y = digits.target

print("original features:", x.shape[1])

scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)

x_train, x_test, y_train, y_test = train_test_split(
    x_scaled, y, test_size=0.2, random_state=42
)

start = time.time()
model1 = SVC(random_state=42)
model1.fit(x_train, y_train)
time1 = round(time.time() - start, 3)
acc1 = round(accuracy_score(y_test, model1.predict(x_test)), 3)

pca = PCA(n_components=0.95)
x_train_pca = pca.fit_transform(x_train)
x_test_pca = pca.transform(x_test)

print("features after pca:", x_train_pca.shape[1])

start = time.time()
model2 = SVC(random_state=42)
model2.fit(x_train_pca, y_train)
time2 = round(time.time() - start, 3)
acc2 = round(accuracy_score(y_test, model2.predict(x_test_pca)), 3)

print("-" * 40)
print("without pca:")
print("  accuracy:", acc1, "| time:", time1, "sec")
print("with pca:")
print("  accuracy:", acc2, "| time:", time2, "sec")
```

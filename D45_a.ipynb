```python
import pandas as pd
import numpy as np

np.random.seed(3)
df = pd.DataFrame({
    'age': np.random.randint(18, 70, 15),
    'income': np.random.exponential(scale=40000, size=15).round(2),
    'hours_worked': np.random.randint(20, 60, 15)
})

print("Original Data:\n", df)

bins = [0, 25, 40, 60, 100]
labels = ['Young', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels)

df['income_log'] = np.log1p(df['income'])

df['income_per_hour'] = df['income'] / df['hours_worked']

df['age_squared'] = df['age'] ** 2

print("\nAfter feature engineering:\n", df)
print("\nSkewness comparison:")
print("Income skew:", df['income'].skew())
print("Log income skew:", df['income_log'].skew())
```

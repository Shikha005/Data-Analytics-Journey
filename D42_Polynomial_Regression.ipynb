```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
x = np.random.uniform(0, 10, 80)
y = 2 * x**2 - 3 * x + 5 + np.random.normal(0, 10, 80)

x_reshape = x.reshape(-1, 1)

linear = LinearRegression()
linear.fit(x_reshape, y)
y_linear = linear.predict(x_reshape)

poly = PolynomialFeatures(degree=2)
x_poly = poly.fit_transform(x_reshape)
poly_model = LinearRegression()
poly_model.fit(x_poly, y)
y_poly = poly_model.predict(x_poly)

print("linear vs polynomial regression")
print("-" * 35)
print("linear r2:", round(r2_score(y, y_linear), 3))
print("polynomial r2:", round(r2_score(y, y_poly), 3))

sorted_idx = np.argsort(x)
plt.figure(figsize=(9, 5))
plt.scatter(x, y, color="blue", alpha=0.4, label="data")
plt.plot(x[sorted_idx], y_linear[sorted_idx],
         color="red", label="linear", linewidth=2)
plt.plot(x[sorted_idx], y_poly[sorted_idx],
         color="green", label="polynomial", linewidth=2)
plt.title("linear vs polynomial regression")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
def collatz_length(n):
    steps = 0
    while n != 1:
        if n % 2 == 0:
            n //= 2
        else:
            n = 3 * n + 1
        steps += 1
    return steps

longest = 0
longest_num = 0
for i in range(1, 10000):
    length = collatz_length(i)
    if length > longest:
        longest = length
        longest_num = i

print(f"Number {longest_num} has the longest Collatz chain: {longest} steps")
```

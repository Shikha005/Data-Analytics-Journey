```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
scores = np.array([50, 55, 65, 70, 75, 78, 85, 90])

slope, intercept, r_value, p_value, std_err = stats.linregress(hours, scores)

print(f"Slope: {slope:.2f}")
print(f"Intercept: {intercept:.2f}")
print(f"R-squared: {r_value**2:.4f}")

predicted_scores = slope * hours + intercept

plt.scatter(hours, scores, label="Actual Data")
plt.plot(hours, predicted_scores, color="red", label="Regression Line")
plt.xlabel("Study Hours")
plt.ylabel("Exam Scores")
plt.title("Linear Regression: Study Hours vs Scores")
plt.legend()
plt.savefig("linear_regression.png")
print("Plot saved as linear_regression.png")
```

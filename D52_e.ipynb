```python
n = int(input("Enter a number: "))

original = n
total = 0

while n > 0:
    digit = n % 10
    n //= 10

    factorial = 1
    i = 1

    while i <= digit:
        factorial *= i
        i += 1

    total += factorial

if total == original:
    print("Strong Number")
else:
    print("Not Strong Number")
```

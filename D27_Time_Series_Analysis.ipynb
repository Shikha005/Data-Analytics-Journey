```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

dates = pd.date_range(start="2024-01-01", periods=12, freq="M")
sales = [200, 220, 210, 250, 270, 260, 300, 320, 310, 340, 360, 380]

df = pd.DataFrame({"Date": dates, "Sales": sales})
df.set_index("Date", inplace=True)

df["Moving_Avg_3"] = df["Sales"].rolling(window=3).mean()

print(df)

# Plot
plt.figure(figsize=(10, 5))
plt.plot(df.index, df["Sales"], label="Actual Sales", marker="o")
plt.plot(df.index, df["Moving_Avg_3"], label="3-Month Moving Average", color="red")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.title("Sales Trend with Moving Average")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("time_series_trend.png")
print("\nPlot saved as time_series_trend.png")
```
